```python
import pandas as pd
import matplotlib.pyplot as plt

# Find columns automatically
sales_col = next((c for c in df.columns if "sales" in c.lower() or "revenue" in c.lower()), None)
profit_col = next((c for c in df.columns if "profit" in c.lower()), None)
quantity_col = next((c for c in df.columns if "quantity" in c.lower() or "qty" in c.lower()), None)
date_col = next((c for c in df.columns if "order date" in c.lower() or "date" in c.lower()), None)
category_col = next((c for c in df.columns if "category" in c.lower()), None)
region_col = next((c for c in df.columns if "region" in c.lower()), None)

print("Detected Columns")
print("Sales:", sales_col)
print("Profit:", profit_col)
print("Quantity:", quantity_col)
print("Date:", date_col)
print("Category:", category_col)
print("Region:", region_col)

if sales_col:
    df[sales_col] = pd.to_numeric(df[sales_col], errors="coerce")

if profit_col:
    df[profit_col] = pd.to_numeric(df[profit_col], errors="coerce")

if quantity_col:
    df[quantity_col] = pd.to_numeric(df[quantity_col], errors="coerce")

print("\n" + "="*50)
print("SALES ANALYSIS")
print("="*50)

if sales_col:
    print("Total Revenue:", f"${df[sales_col].sum():,.2f}")

if profit_col:
    print("Total Profit:", f"${df[profit_col].sum():,.2f}")

if quantity_col:
    print("Total Quantity:", f"{df[quantity_col].sum():,.0f}")

# Category chart
if category_col and sales_col:
    category_data = df.groupby(category_col)[sales_col].sum().sort_values(ascending=False)

    plt.figure(figsize=(8,5))
    category_data.plot(kind="bar")
    plt.title("Revenue by Category")
    plt.xlabel("Category")
    plt.ylabel("Revenue")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

# Region chart
if region_col and sales_col:
    region_data = df.groupby(region_col)[sales_col].sum().sort_values(ascending=False)

    plt.figure(figsize=(8,5))
    region_data.plot(kind="bar")
    plt.title("Revenue by Region")
    plt.xlabel("Region")
    plt.ylabel("Revenue")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

print("\nAnalysis completed.")
```

```python
# Advanced Sales & Revenue Analysis

print("=" * 60)
print("ADVANCED SALES & REVENUE ANALYSIS")
print("=" * 60)

# Date conversion
if date_col:
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

# Monthly Revenue Trend
if date_col and sales_col:
    monthly = df.dropna(subset=[date_col]).groupby(
        df[date_col].dt.to_period("M")
    )[sales_col].sum()

    plt.figure(figsize=(12,5))
    plt.plot(monthly.index.astype(str), monthly.values, marker="o")
    plt.title("Monthly Revenue Trend")
    plt.xlabel("Month")
    plt.ylabel("Revenue")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Profit by Category
if category_col and profit_col:
    category_profit = df.groupby(category_col)[profit_col].sum().sort_values(ascending=False)

    plt.figure(figsize=(8,5))
    category_profit.plot(kind="bar")
    plt.title("Profit by Category")
    plt.xlabel("Category")
    plt.ylabel("Profit")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

# Revenue by Region
if region_col and sales_col:
    region_revenue = df.groupby(region_col)[sales_col].sum().sort_values(ascending=False)

    plt.figure(figsize=(8,5))
    region_revenue.plot(kind="bar")
    plt.title("Revenue by Region")
    plt.xlabel("Region")
    plt.ylabel("Revenue")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

print("Advanced analysis completed successfully.")
```

```python
print("=" * 60)
print("TOP PRODUCTS & PROFITABILITY ANALYSIS")
print("=" * 60)

# Detect product and discount columns
product_col = next(
    (c for c in df.columns if "product" in c.lower() and "name" in c.lower()),
    next((c for c in df.columns if "product" in c.lower()), None)
)

discount_col = next(
    (c for c in df.columns if "discount" in c.lower()),
    None
)

# Top 10 Products
if product_col and sales_col:
    top_products = (
        df.groupby(product_col)[sales_col]
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    print("\nTop 10 Products by Revenue:")
    print(top_products.round(2))

    plt.figure(figsize=(10,6))
    top_products.sort_values().plot(kind="barh")
    plt.title("Top 10 Products by Revenue")
    plt.xlabel("Revenue")
    plt.ylabel("Product")
    plt.tight_layout()
    plt.show()

# Profit Margin
if sales_col and profit_col:
    total_sales = df[sales_col].sum()
    total_profit = df[profit_col].sum()

    profit_margin = (total_profit / total_sales) * 100

    print(f"\nTotal Revenue: ${total_sales:,.2f}")
    print(f"Total Profit: ${total_profit:,.2f}")
    print(f"Overall Profit Margin: {profit_margin:.2f}%")

# Discount Analysis
if discount_col and sales_col and profit_col:
    df[discount_col] = pd.to_numeric(
        df[discount_col], errors="coerce"
    )

    discount_analysis = df.groupby(discount_col)[
        [sales_col, profit_col]
    ].sum()

    print("\nDiscount vs Sales & Profit:")
    display(discount_analysis.round(2))

print("\nCell 4 completed successfully.")
```

```python
print("=" * 60)
print("BUSINESS INSIGHTS & RECOMMENDATIONS")
print("=" * 60)

# Category insight
if category_col and sales_col:
    category_sales = df.groupby(category_col)[sales_col].sum()
    best_category = category_sales.idxmax()
    best_category_value = category_sales.max()

    print(f"\n1. Top Revenue Category: {best_category}")
    print(f"   Revenue: ${best_category_value:,.2f}")

# Region insight
if region_col and sales_col:
    region_sales = df.groupby(region_col)[sales_col].sum()
    best_region = region_sales.idxmax()
    best_region_value = region_sales.max()

    print(f"\n2. Top Revenue Region: {best_region}")
    print(f"   Revenue: ${best_region_value:,.2f}")

# Profit insight
if category_col and profit_col:
    category_profit = df.groupby(category_col)[profit_col].sum()
    best_profit_category = category_profit.idxmax()

    print(f"\n3. Most Profitable Category: {best_profit_category}")
    print(f"   Profit: ${category_profit.max():,.2f}")

# Product insight
if product_col and sales_col:
    product_sales = df.groupby(product_col)[sales_col].sum()
    top_product = product_sales.idxmax()

    print(f"\n4. Highest Revenue Product: {top_product}")
    print(f"   Revenue: ${product_sales.max():,.2f}")

# Overall metrics
if sales_col and profit_col:
    revenue = df[sales_col].sum()
    profit = df[profit_col].sum()
    margin = (profit / revenue) * 100

    print(f"\n5. Overall Profit Margin: {margin:.2f}%")

print("\n" + "=" * 60)
print("RECOMMENDATIONS")
print("=" * 60)

print("""
• Focus marketing efforts on high-performing categories and regions.
• Monitor products with high sales but relatively low profit.
• Analyze discount levels to avoid unnecessary profit reduction.
• Maintain inventory availability for consistently high-performing products.
• Use monthly revenue trends for sales forecasting and planning.
""")

print("\nSales & Revenue Analysis completed successfully.")
```
